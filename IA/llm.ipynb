```python
import ollama

def calculer_score_rgesn(evaluations_ia):
    """
    Calcule le score de maturité en fonction des retours de l'IA.
    evaluations_ia : dictionnaire au format {"Nom du critère": "Validé" ou "Non Validé"}
    """
    # Sécurité pour éviter de diviser par zéro si le dictionnaire est vide
    if not evaluations_ia:
        return 0.0
    
    total_criteres = len(evaluations_ia)
    criteres_valides = 0
    
    # On compte combien de critères ont reçu le feu vert
    for critere, statut in evaluations_ia.items():
        # .strip().lower() permet d'éviter les bugs si l'IA met un espace ou une majuscule en trop
        if statut.strip().lower() in ["validé", "valide"]:
            criteres_valides += 1
            
    # Calcul du score sur 100
    score_final = (criteres_valides / total_criteres) * 100
    return round(score_final, 2)


# =====================================================================
# ZONE DE PROMPTING ET D'INJECTION DU CONTEXTE
# =====================================================================

# 1. C'est ici que tu injecteras les données récupérées depuis ton interface 
# (par exemple : le code HTML collé par l'utilisateur, ses réponses au chatbot, etc.)
contexte_entreprise = """
[... ZONE VIDE : INSERER LE CONTEXTE DE L'ENTREPRISE ICI ...]
"""

# 2. Le System Prompt : On verrouille le comportement de l'IA
# L'astuce ici est de forcer l'IA à répondre avec un format très strict pour que 
# Python puisse lire sa réponse facilement.
system_prompt = f"""
Tu es un auditeur expert en écoconception numérique basé sur le RGESN.
Voici les informations fournies par l'entreprise : 
{contexte_entreprise}

Ta mission est d'évaluer ces données par rapport aux critères suivants :
1. Stratégie d'hébergement green.
2. Optimisation et compression des images.
3. Absence de vidéos en lecture automatique (autoplay).
[... ZONE VIDE : INSERER LE RESTE DES CRITÈRES RGESN ICI ...]

Consigne stricte : Pour CHAQUE critère, tu dois répondre UNIQUEMENT par le nom du critère suivi de "Validé" ou "Non Validé". Ne fais aucune phrase.
"""

# =====================================================================
# EXECUTION (Exemple de flux de travail)
# =====================================================================

if __name__ == "__main__":
    print("⏳ Lancement de l'analyse IA en local...")
    
    # ÉTAPE 1 : Appel à l'IA (Commenté car tu devras l'adapter à ton flux)
    """
    reponse = ollama.chat(model='mistral', messages=[
        {'role': 'system', 'content': system_prompt}
    ])
    reponse_brute_ia = reponse['message']['content']
    # Ici, tu devras écrire un petit bout de code pour transformer le texte brut de l'IA
    # en un dictionnaire Python (resultats_extraits).
    """
    
    # ÉTAPE 2 : Simulation de la réponse de l'IA (une fois formatée en dictionnaire)
    resultats_extraits = {
        "Stratégie d'hébergement green": "Non Validé",
        "Optimisation et compression des images": "Validé",
        "Absence de vidéos en lecture automatique": "Validé"
    }
    
    # ÉTAPE 3 : Python prend le relais pour les mathématiques
    score = calculer_score_rgesn(resultats_extraits)
    
    print("\n📊 Résultats bruts extraits de l'IA :")
    for critere, note in resultats_extraits.items():
        print(f" - {critere} : {note}")
        
    print(f"\n🏆 Score de maturité RGESN calculé : {score}/100")
```
